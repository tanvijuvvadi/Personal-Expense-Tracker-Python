```python
# Cleaning the data and making it ready for analysis

# Step 1: Convert the date column from text to real dates
df['date'] = pd.to_datetime(df['date'])

# Step 2: Add a month column (example: "January 2024")
df['month_name'] = df['date'].dt.strftime('%B %Y')

# Step 3: Add a month column for grouping (example: "2024-01")
df['month'] = df['date'].dt.strftime('%Y-%m')

# Step 4: Make sure amount is a number
df['amount'] = pd.to_numeric(df['amount'], errors='coerce')

# Step 5: Remove any empty/broken rows
df = df.dropna()

# Show the cleaned data
print("Data cleaned successfully!")
print(f"Total records after cleaning: {len(df)}")
print()
print(df[['date', 'category', 'amount', 'payment_method', 'month_name']].head(10))
```

```python
# Finding out how much was spent in each category

# Add up all amounts for each category
category_spending = df.groupby('category')['amount'].sum().sort_values(ascending=False)

# Calculate total spending
total_spending = df['amount'].sum()

# Calculate average spending per day
avg_daily = df.groupby('date')['amount'].sum().mean()

# Find the highest spending category
top_category = category_spending.index[0]
top_amount = category_spending.values[0]

# Print everything nicely
print("=" * 45)
print("       CATEGORY-WISE SPENDING ANALYSIS")
print("=" * 45)
print()
print(f"Total Money Spent    : Rs. {total_spending:,.2f}")
print(f"Average Daily Spend  : Rs. {avg_daily:,.2f}")
print(f"Highest Category     : {top_category} (Rs. {top_amount:,.2f})")
print()
print("-" * 45)
print(f"{'Category':<18} {'Amount':>10} {'% of Total':>12}")
print("-" * 45)

for category, amount in category_spending.items():
    percentage = (amount / total_spending) * 100
    print(f"{category:<18} Rs.{amount:>7,.0f}   {percentage:>8.1f}%")

print("-" * 45)
print(f"{'TOTAL':<18} Rs.{total_spending:>7,.0f}   {'100.0%':>9}")
print("=" * 45)
```

```python
# Finding out how much was spent each month

# Add up all amounts for each month
monthly_spending = df.groupby('month_name')['amount'].sum()

# Sort by month order (Jan → Feb → Mar)
month_order = ['January 2024', 'February 2024', 'March 2024']
monthly_spending = monthly_spending.reindex(month_order)

# Find highest and lowest spending month
best_month = monthly_spending.idxmin()
worst_month = monthly_spending.idxmax()

# Print monthly analysis
print("=" * 45)
print("         MONTHLY SPENDING ANALYSIS")
print("=" * 45)
print()

for month, amount in monthly_spending.items():
    # Make a simple bar using dashes
    bar_length = int(amount / 200)
    bar = "█" * bar_length
    print(f"{month:<20} Rs.{amount:>7,.0f}")
    print(f"                     {bar}")
    print()

print("-" * 45)
print(f"Lowest spending month  : {best_month}")
print(f"Highest spending month : {worst_month}")
print()

# Month on month change
amounts = monthly_spending.values
for i in range(1, len(amounts)):
    change = amounts[i] - amounts[i-1]
    direction = "up" if change > 0 else "down"
    print(f"{month_order[i-1]} to {month_order[i]}: Rs.{abs(change):,.0f} {direction}")

print("=" * 45)
```

```python
# Creating a colorful bar chart showing spending by category

# Set the chart size and style
plt.figure(figsize=(10, 6))
sns.set_style("whitegrid")

# Define colors for each bar
colors = ['#FF6B6B', '#4ECDC4', '#45B7D1', '#96CEB4', '#FFEAA7', '#DDA0DD', '#98D8C8']

# Create the bar chart
bars = plt.bar(category_spending.index,
               category_spending.values,
               color=colors,
               edgecolor='white',
               linewidth=1.5,
               width=0.6)

# Add amount labels on top of each bar
for bar, value in zip(bars, category_spending.values):
    plt.text(
        bar.get_x() + bar.get_width() / 2,  # center of bar
        bar.get_height() + 50,               # just above bar
        f'Rs.{value:,.0f}',                  # the amount text
        ha='center',
        va='bottom',
        fontsize=9,
        fontweight='bold',
        color='#333333'
    )

# Add titles and labels
plt.title('Category-wise Expense Breakdown',
          fontsize=16, fontweight='bold', pad=20, color='#2C3E50')
plt.xlabel('Category', fontsize=12, color='#2C3E50')
plt.ylabel('Total Amount Spent (Rs.)', fontsize=12, color='#2C3E50')
plt.xticks(rotation=30, ha='right', fontsize=10)
plt.yticks(fontsize=10)

# Add a horizontal line showing average spending per category
avg_per_cat = category_spending.mean()
plt.axhline(y=avg_per_cat, color='red', linestyle='--', linewidth=1.5, alpha=0.7)
plt.text(len(category_spending)-0.5, avg_per_cat + 50,
         f'Average: Rs.{avg_per_cat:,.0f}',
         color='red', fontsize=9, ha='right')

plt.tight_layout()
plt.savefig('chart1_category_bar.png', dpi=150, bbox_inches='tight')
plt.show()

print("Chart 1 saved as: chart1_category_bar.png")
```

```python
# Creating a line chart showing how spending changed month by month

plt.figure(figsize=(10, 6))
sns.set_style("whitegrid")

# Get months and amounts as lists
months = list(monthly_spending.index)
amounts = list(monthly_spending.values)

# Draw the line
plt.plot(months, amounts,
         marker='o',           # circle at each point
         linewidth=2.5,
         color='#2196F3',
         markersize=12,
         markerfacecolor='white',
         markeredgewidth=2.5,
         markeredgecolor='#2196F3')

# Add amount labels above each point
for x, y in zip(months, amounts):
    plt.text(x, y + 100,
             f'Rs.{y:,.0f}',
             ha='center',
             fontsize=11,
             fontweight='bold',
             color='#2196F3')

# Shade the area under the line
plt.fill_between(months, amounts,
                 alpha=0.1,
                 color='#2196F3')

# Add titles and labels
plt.title('Monthly Spending Trend',
          fontsize=16, fontweight='bold', pad=20, color='#2C3E50')
plt.xlabel('Month', fontsize=12, color='#2C3E50')
plt.ylabel('Total Amount Spent (Rs.)', fontsize=12, color='#2C3E50')
plt.xticks(fontsize=11)
plt.yticks(fontsize=10)

# Add a note if spending went up
if amounts[-1] > amounts[0]:
    plt.text(months[-1], amounts[-1] - 400,
             'Spending is increasing!',
             ha='right', fontsize=10,
             color='red', style='italic')

plt.tight_layout()
plt.savefig('chart2_monthly_trend.png', dpi=150, bbox_inches='tight')
plt.show()

print("Chart 2 saved as: chart2_monthly_trend.png")
```

```python
# Creating a pie chart showing how payments were made

# Add up spending by payment method
payment_spending = df.groupby('payment_method')['amount'].sum()

plt.figure(figsize=(9, 7))

# Define colors for each slice
colors = ['#FF6B6B', '#4ECDC4', '#45B7D1', '#FFEAA7']

# Create the pie chart (donut style)
wedges, texts, autotexts = plt.pie(
    payment_spending.values,
    labels=payment_spending.index,
    autopct='%1.1f%%',          # show percentage on each slice
    startangle=140,
    colors=colors,
    wedgeprops=dict(width=0.6,  # makes it a donut shape
                    edgecolor='white',
                    linewidth=2),
    pctdistance=0.75            # where the % text sits
)

# Make the label text bigger and bold
for text in texts:
    text.set_fontsize(12)
    text.set_fontweight('bold')

# Make the percentage text bigger
for autotext in autotexts:
    autotext.set_fontsize(11)
    autotext.set_fontweight('bold')
    autotext.set_color('white')

# Add total in the center of the donut
total = payment_spending.sum()
plt.text(0, 0, f'Total\nRs.{total:,.0f}',
         ha='center', va='center',
         fontsize=13, fontweight='bold',
         color='#2C3E50')

# Add title
plt.title('Spending by Payment Method',
          fontsize=16, fontweight='bold', pad=20, color='#2C3E50')

# Add a legend
plt.legend(
    [f'{method}: Rs.{amt:,.0f}' for method, amt in payment_spending.items()],
    loc='lower center',
    bbox_to_anchor=(0.5, -0.08),
    ncol=2,
    fontsize=10,
    frameon=False
)

plt.tight_layout()
plt.savefig('chart3_payment_pie.png', dpi=150, bbox_inches='tight')
plt.show()

print("Chart 3 saved as: chart3_payment_pie.png")
print()
print("Payment Method Breakdown:")
for method, amount in payment_spending.items():
    pct = (amount / total) * 100
    print(f"  {method:<10} : Rs.{amount:>7,.0f}  ({pct:.1f}%)")
```

```python
# Creating a bar chart showing spending on each individual day

# Add up spending for each date
daily_spending = df.groupby('date')['amount'].sum().reset_index()
daily_spending.columns = ['date', 'amount']

# Find the highest spending day
max_day = daily_spending.loc[daily_spending['amount'].idxmax()]
min_day = daily_spending.loc[daily_spending['amount'].idxmin()]

plt.figure(figsize=(14, 6))
sns.set_style("whitegrid")

# Give each bar a color based on amount
# High spending = red, Low spending = green
colors = []
for amount in daily_spending['amount']:
    if amount >= 1500:
        colors.append('#FF6B6B')   # red = high spending
    elif amount >= 500:
        colors.append('#FFEAA7')   # yellow = medium spending
    else:
        colors.append('#96CEB4')   # green = low spending

# Draw the bars
bars = plt.bar(daily_spending['date'],
               daily_spending['amount'],
               color=colors,
               edgecolor='white',
               linewidth=0.8,
               width=0.6)

# Highlight the highest spending day with a star label
plt.annotate(
    f"Highest!\nRs.{max_day['amount']:,.0f}",
    xy=(max_day['date'], max_day['amount']),
    xytext=(max_day['date'], max_day['amount'] + 150),
    ha='center',
    fontsize=9,
    fontweight='bold',
    color='#FF6B6B',
    arrowprops=dict(arrowstyle='->', color='#FF6B6B')
)

# Highlight the lowest spending day
plt.annotate(
    f"Lowest!\nRs.{min_day['amount']:,.0f}",
    xy=(min_day['date'], min_day['amount']),
    xytext=(min_day['date'], min_day['amount'] + 300),
    ha='center',
    fontsize=9,
    fontweight='bold',
    color='#27AE60',
    arrowprops=dict(arrowstyle='->', color='#27AE60')
)

# Add titles and labels
plt.title('Daily Spending Trend (Jan - Mar 2024)',
          fontsize=16, fontweight='bold', pad=20, color='#2C3E50')
plt.xlabel('Date', fontsize=12, color='#2C3E50')
plt.ylabel('Amount Spent (Rs.)', fontsize=12, color='#2C3E50')
plt.xticks(rotation=45, ha='right', fontsize=8)
plt.yticks(fontsize=10)

# Add color legend
from matplotlib.patches import Patch
legend_items = [
    Patch(facecolor='#FF6B6B', label='High spending (Rs.1500+)'),
    Patch(facecolor='#FFEAA7', label='Medium spending (Rs.500-1499)'),
    Patch(facecolor='#96CEB4', label='Low spending (below Rs.500)')
]
plt.legend(handles=legend_items, loc='upper left', fontsize=9, frameon=True)

plt.tight_layout()
plt.savefig('chart4_daily_trend.png', dpi=150, bbox_inches='tight')
plt.show()

print("Chart 4 saved as: chart4_daily_trend.png")
print()
print(f"Highest spending day : {max_day['date'].strftime('%d %B %Y')} — Rs.{max_day['amount']:,.0f}")
print(f"Lowest spending day  : {min_day['date'].strftime('%d %B %Y')} — Rs.{min_day['amount']:,.0f}")
```

```python
# Generating the final summary report

now = datetime.now().strftime('%d %B %Y %H:%M')

report = f"""
╔══════════════════════════════════════════════╗
       PERSONAL EXPENSE TRACKER REPORT
       Generated on: {now}
╚══════════════════════════════════════════════╝

OVERVIEW
---------
Total Records Analyzed  : {len(df)}
Total Money Spent       : Rs. {total_spending:,.2f}
Average Daily Spending  : Rs. {avg_daily:,.2f}
Highest Spending Area   : {top_category} (Rs. {top_amount:,.2f})

CATEGORY-WISE BREAKDOWN
------------------------
"""

for category, amount in category_spending.items():
    pct = (amount / total_spending) * 100
    bar = "█" * int(pct / 2)
    report += f"{category:<18} Rs.{amount:>7,.0f}  {pct:>5.1f}%  {bar}\n"

report += f"""
MONTHLY BREAKDOWN
------------------
"""

for month, amount in monthly_spending.items():
    pct = (amount / total_spending) * 100
    report += f"{month:<22} Rs.{amount:>7,.0f}  {pct:>5.1f}%\n"

report += f"""
PAYMENT METHOD BREAKDOWN
-------------------------
"""

for method, amount in payment_spending.items():
    pct = (amount / total_spending) * 100
    report += f"{method:<12} Rs.{amount:>7,.0f}  {pct:>5.1f}%\n"

report += f"""
KEY INSIGHTS
-------------
1. You spent the most on {top_category} — consider reducing it
2. Your spending went {'UP' if amounts[-1] > amounts[0] else 'DOWN'} from {months[0]} to {months[-1]}
3. Highest single day spend : Rs.{max_day['amount']:,.0f} on {max_day['date'].strftime('%d %B %Y')}
4. Lowest single day spend  : Rs.{min_day['amount']:,.0f} on {min_day['date'].strftime('%d %B %Y')}
5. Most used payment method : {payment_spending.idxmax()}

╔══════════════════════════════════════════════╗
   Project by: YOUR NAME | Python Expense Tracker
╚══════════════════════════════════════════════╝
"""

# Print the report
print(report)

# Save the report as a text file
with open('expense_report.txt', 'w') as f:
    f.write(report)

# Save category summary as CSV
category_summary = category_spending.reset_index()
category_summary.columns = ['category', 'total_spent']
category_summary['percentage'] = (category_summary['total_spent'] / total_spending * 100).round(1)
category_summary.to_csv('category_summary.csv', index=False)

# Save monthly summary as CSV
monthly_summary = monthly_spending.reset_index()
monthly_summary.columns = ['month', 'total_spent']
monthly_summary.to_csv('monthly_summary.csv', index=False)

print("Files saved successfully!")
print()
print("  expense_report.txt     - your full text report")
print("  category_summary.csv   - category data")
print("  monthly_summary.csv    - monthly data")
```
